## 1. Total Revenue (for successfully delivered orders)

```python
delivered_orders = df[df["order_status"] == "delivered"]
total_revenue = (delivered_orders["price"] + delivered_orders["freight_value"]).sum()
print("Total_Revenue = ",total_revenue)
```

## 2. Expected Revenue (all approved orders) 

```python
approved_orders = df[df["order_status"].isin(["delivered", "shipped", "processing", "approved"])]
expected_revenue = approved_orders["payment_value"].sum()
print("Expected_Revenue = ",expected_revenue)
```

## Bar chart for revenue comparison

```python
plt.figure(figsize=(8, 5))
plt.bar(["Total Revenue", "Expected Revenue"], [total_revenue, expected_revenue], color=['blue', 'orange'])
plt.ylabel("Revenue (in currency units)")
plt.title("Total vs Expected Revenue")
plt.show()
```

## 3. Canceled Orders Count

```python
canceled_orders_count = df[df["order_status"] == "canceled"].shape[0]
print("canceled_orders_count = ", canceled_orders_count)
```

## 4. Late Deliveries Count

```python
late_deliveries_count = df[df["order_delivered_customer_date"] > df["order_estimated_delivery_date"]].shape[0]
```

```python
print("Late_deliveries_count = ",late_deliveries_count )
```
